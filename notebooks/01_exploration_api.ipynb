```python
import requests

url = "https://data.economie.gouv.fr/api/explore/v2.1/catalog/datasets/prix-des-carburants-en-france-flux-instantane-v2/records"
params = {"limit": 5}

reponse = requests.get(url, params=params)
reponse.raise_for_status()
data = reponse.json()

print("Nombre total de stations :", data["total_count"])

premiere_station = data["results"][0]
print(list(premiere_station.keys()))

```
